# Disjoint Set

## Union by rank

Union(u,v)
- finr the ultimate parent of u and v: pu and pv
- find the rank of those parents of pu and pv
- connect smaller rank to larger rank always

A Disjoint Set (also called Union-Find) is a data structure used in graphs to keep track of connected components efficiently.

Think of it like this:
You have a bunch of nodes, and initially each node is in its own group. As you connect nodes with edges, some groups merge. The disjoint set helps you quickly answer:

- “Are these two nodes in the same group?”
- “Merge these two groups.”

for union -> TC: O(4alpha)
for findParent -> TC: O(4alpha)
